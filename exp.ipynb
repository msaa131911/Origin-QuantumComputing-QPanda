```python
from pyqpanda3.core import QCircuit, QProg, H, CNOT, measure, CPUQVM

circuit = QCircuit(2)
circuit << H(0)
circuit << CNOT(0, 1)

prog = QProg()
prog << circuit
prog << measure(0, 0) << measure(1, 1)

machine = CPUQVM()
machine.run(prog, 1000)
print(machine.result().get_counts())
```
